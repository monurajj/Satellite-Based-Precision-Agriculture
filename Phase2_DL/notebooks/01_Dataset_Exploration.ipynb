# Notebook 01: Dataset Exploration

**Runs on:** Local machine (no GPU needed)

This notebook downloads and explores the EuroSAT dataset:
- Dataset overview and class distribution
- Sample image visualization per class
- Data augmentation demonstration
- Train/Val/Test split creation
- Save split indices for use in all subsequent notebooks

```python
import os
import sys
import json
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

import torch
from torch.utils.data import DataLoader
from torchvision import transforms
from torchvision.datasets import EuroSAT
from sklearn.model_selection import train_test_split

# Reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

# Paths
DATA_DIR = os.path.join(os.path.dirname(os.getcwd()), 'data')
RESULTS_DIR = os.path.join(os.path.dirname(os.getcwd()), 'experiments', 'results')
FIGURES_DIR = os.path.join(RESULTS_DIR, 'figures')
os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(FIGURES_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

print(f"Data directory: {DATA_DIR}")
print(f"Results directory: {RESULTS_DIR}")
```

## 1. Download EuroSAT Dataset

**EuroSAT** (Helber et al., 2019, IEEE JSTARS)
- 27,000 Sentinel-2 satellite image patches (64x64 pixels)
- 10 land use/land cover classes
- Same satellite platform as Phase 1 (Sentinel-2)
- Agriculture-relevant classes: AnnualCrop, PermanentCrop, Pasture

```python
# Download EuroSAT RGB dataset
raw_dataset = EuroSAT(root=DATA_DIR, download=True)

CLASS_NAMES = [
    'AnnualCrop', 'Forest', 'HerbaceousVegetation', 'Highway',
    'Industrial', 'Pasture', 'PermanentCrop', 'Residential',
    'River', 'SeaLake'
]

print(f"Total images: {len(raw_dataset)}")
print(f"Number of classes: {len(CLASS_NAMES)}")
print(f"Image size: {raw_dataset[0][0].size}")
print(f"\nClasses:")
for i, name in enumerate(CLASS_NAMES):
    print(f"  {i}: {name}")
```

## 2. Class Distribution Analysis

```python
# Compute class distribution
all_labels = [raw_dataset[i][1] for i in range(len(raw_dataset))]
label_counts = Counter(all_labels)

print("Class Distribution:")
print("-" * 50)
for cls_idx in sorted(label_counts.keys()):
    name = CLASS_NAMES[cls_idx]
    count = label_counts[cls_idx]
    pct = count / len(all_labels) * 100
    ag_flag = " [AGRICULTURE]" if name in ['AnnualCrop', 'PermanentCrop', 'Pasture'] else ""
    print(f"  {name:<25} {count:>5} ({pct:5.1f}%){ag_flag}")

# Check for class imbalance
min_count = min(label_counts.values())
max_count = max(label_counts.values())
imbalance_ratio = max_count / min_count
print(f"\nImbalance ratio (max/min): {imbalance_ratio:.2f}")
print(f"Minimum class size: {min_count}")
print(f"Maximum class size: {max_count}")
```

```python
# Plot class distribution
fig, ax = plt.subplots(figsize=(12, 5))
counts = [label_counts[i] for i in range(len(CLASS_NAMES))]
colors = ['#2ecc71' if name in ['AnnualCrop', 'PermanentCrop', 'Pasture'] else '#3498db'
          for name in CLASS_NAMES]

bars = ax.bar(CLASS_NAMES, counts, color=colors, edgecolor='white', linewidth=0.5)
ax.set_xlabel('Land Cover Class', fontsize=12)
ax.set_ylabel('Number of Images', fontsize=12)
ax.set_title('EuroSAT Class Distribution (Green = Agriculture Classes)', fontsize=14)
ax.axhline(y=np.mean(counts), color='red', linestyle='--', alpha=0.5, label=f'Mean: {np.mean(counts):.0f}')
ax.legend(fontsize=10)

# Add count labels
for bar, count in zip(bars, counts):
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 30,
            str(count), ha='center', va='bottom', fontsize=9)

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'class_distribution.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## 3. Sample Image Visualization

```python
# Show 3 samples per class (30 total)
fig, axes = plt.subplots(len(CLASS_NAMES), 3, figsize=(9, 30))

# Collect samples per class
class_samples = {i: [] for i in range(len(CLASS_NAMES))}
for i in range(len(raw_dataset)):
    img, label = raw_dataset[i]
    if len(class_samples[label]) < 3:
        class_samples[label].append(img)
    if all(len(v) >= 3 for v in class_samples.values()):
        break

for cls_idx in range(len(CLASS_NAMES)):
    for col in range(3):
        axes[cls_idx][col].imshow(class_samples[cls_idx][col])
        axes[cls_idx][col].axis('off')
        if col == 0:
            axes[cls_idx][col].set_ylabel(CLASS_NAMES[cls_idx], fontsize=10,
                                          rotation=0, labelpad=120, va='center')

axes[0][1].set_title('Sample Sentinel-2 Images', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'sample_images.png'), dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Pixel intensity statistics per class
to_tensor = transforms.ToTensor()
class_means = {}

# Sample 100 images per class for stats
class_pixels = {i: [] for i in range(len(CLASS_NAMES))}
for i in range(len(raw_dataset)):
    img, label = raw_dataset[i]
    if len(class_pixels[label]) < 100:
        tensor = to_tensor(img)
        class_pixels[label].append(tensor.mean(dim=(1, 2)).numpy())  # mean per channel
    if all(len(v) >= 100 for v in class_pixels.values()):
        break

# Plot per-channel means
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
channel_names = ['Red', 'Green', 'Blue']
channel_colors = ['#e74c3c', '#2ecc71', '#3498db']

for ch in range(3):
    means_per_class = []
    for cls_idx in range(len(CLASS_NAMES)):
        vals = [p[ch] for p in class_pixels[cls_idx]]
        means_per_class.append(np.mean(vals))

    axes[ch].bar(CLASS_NAMES, means_per_class, color=channel_colors[ch], alpha=0.7)
    axes[ch].set_title(f'{channel_names[ch]} Channel Mean', fontsize=12, fontweight='bold')
    axes[ch].set_ylabel('Mean Intensity')
    axes[ch].tick_params(axis='x', rotation=45)

plt.suptitle('Per-Channel Pixel Statistics by Class', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'pixel_statistics.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## 4. Data Augmentation Demonstration

Satellite-appropriate augmentations:
- **H/V Flips**: Satellite images have no canonical orientation
- **90-degree Rotations**: Same reason as flips
- **Brightness/Contrast Jitter**: Simulates atmospheric variation
- **NO perspective transforms**: Would distort geospatial meaning

```python
# Define augmentation transforms
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomApply([
        transforms.RandomRotation(degrees=[90, 90]),
    ], p=0.5),
    transforms.RandomApply([
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1),
    ], p=0.3),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

# Visualize augmented versions of one sample
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
sample_img, sample_label = raw_dataset[0]

axes[0][0].imshow(sample_img)
axes[0][0].set_title('Original', fontsize=12, fontweight='bold')
axes[0][0].axis('off')

for i in range(1, 10):
    row, col = divmod(i, 5)
    augmented = train_transform(sample_img)
    # Denormalize for display
    img_display = augmented.clone()
    for c in range(3):
        img_display[c] = img_display[c] * IMAGENET_STD[c] + IMAGENET_MEAN[c]
    img_display = img_display.clamp(0, 1)
    axes[row][col].imshow(img_display.permute(1, 2, 0).numpy())
    axes[row][col].set_title(f'Augmented #{i}', fontsize=11)
    axes[row][col].axis('off')

plt.suptitle(f'Data Augmentation Examples ({CLASS_NAMES[sample_label]})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, 'augmentation_examples.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Create and Save Train/Val/Test Split

**Strategy:** 70% train / 15% validation / 15% test, stratified by class.

Split indices are saved to `data/split_indices.json` so all notebooks use the same split.

```python
# Stratified split
all_indices = list(range(len(raw_dataset)))
all_labels = [raw_dataset[i][1] for i in all_indices]

# 70% train, 30% temp
train_idx, temp_idx, train_labels, temp_labels = train_test_split(
    all_indices, all_labels, test_size=0.30, random_state=SEED, stratify=all_labels
)

# 50/50 of temp = 15% val, 15% test
val_idx, test_idx, val_labels, test_labels = train_test_split(
    temp_idx, temp_labels, test_size=0.50, random_state=SEED, stratify=temp_labels
)

print(f"Train: {len(train_idx)} ({len(train_idx)/len(all_indices)*100:.0f}%)")
print(f"Val:   {len(val_idx)} ({len(val_idx)/len(all_indices)*100:.0f}%)")
print(f"Test:  {len(test_idx)} ({len(test_idx)/len(all_indices)*100:.0f}%)")

# Verify stratification
print("\nClass distribution per split:")
for split_name, split_labels in [('Train', train_labels), ('Val', val_labels), ('Test', test_labels)]:
    dist = Counter(split_labels)
    print(f"  {split_name}: {dict(sorted(dist.items()))}")
```

```python
# Save split indices for all other notebooks to use
split_data = {
    'train_idx': train_idx,
    'val_idx': val_idx,
    'test_idx': test_idx,
    'class_names': CLASS_NAMES,
    'seed': SEED,
    'imagenet_mean': IMAGENET_MEAN,
    'imagenet_std': IMAGENET_STD,
}

split_path = os.path.join(DATA_DIR, 'split_indices.json')
with open(split_path, 'w') as f:
    json.dump(split_data, f, indent=2)

print(f"Split indices saved to: {split_path}")
print(f"\nAll subsequent notebooks will load this file to ensure consistent splits.")
```

```python
# Compute class weights for weighted loss (saved for training notebook)
train_label_counts = Counter(train_labels)
total_train = len(train_labels)
class_weights = [total_train / (len(CLASS_NAMES) * train_label_counts[i]) for i in range(len(CLASS_NAMES))]

# Add to split data
split_data['class_weights'] = class_weights
with open(split_path, 'w') as f:
    json.dump(split_data, f, indent=2)

print("Class weights (inverse frequency):")
for i, name in enumerate(CLASS_NAMES):
    print(f"  {name:<25} {class_weights[i]:.3f}")

print(f"\nUpdated {split_path} with class weights.")
```
